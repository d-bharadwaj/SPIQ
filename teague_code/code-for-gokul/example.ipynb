```python
def build_max_cut_hypergraph_paulis_from_dict(graph_dict: dict) -> list[tuple[str, float]]:
    n=8
    """Convert the hypergraph dictionary to Pauli list."""
    pauli_list = []

    for edge, weight in graph_dict.items():

        if len(edge) > 1:

            # Initialize Pauli string with identity operators for all nodes
            paulis = ["I"] * (n)  # Ensure the list is large enough to handle the largest node index
            
            # Apply "Z" to each node involved in the edge
            for node in edge:
                paulis[node] = "Z"

            # Add the Pauli string to the list, ensuring the order of qubits is reversed
            pauli_list.append(("".join(paulis)[::-1], weight))

    return pauli_list

max_cut_paulis = build_max_cut_hypergraph_paulis_from_dict(pubo)
```

```python
cost_hamiltonian = SparsePauliOp.from_list(max_cut_paulis)
cost_hamiltonian
```

```python
# Solve with classical Eigensolver for comparison
from qiskit_algorithms import NumPyMinimumEigensolver

eigensolver = NumPyMinimumEigensolver()
exact_solution = eigensolver.compute_minimum_eigenvalue(cost_hamiltonian).eigenvalue.real
print("Exact Energy from Eigensolver:", exact_solution)
```
