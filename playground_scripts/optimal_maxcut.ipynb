```python
from sage.all import Graph

# Convert rustworkx graph to SageMath Graph
# Pass the edge list as (u, v, weight) tuples
def compute_optimal_max_cut(graph):
    sage_graph = Graph()
    for u, v, weight in graph.weighted_edge_list():
        sage_graph.add_edge(u, v, weight)

    # Compute the Max-Cut
    return sage_graph.max_cut(use_edge_labels=True)

# Call the function with the graph
compute_optimal_max_cut(graph)

# # Print results
# print("Max-Cut Value:", max_cut_value)
# print("Max-Cut Partition:", max_cut_partition)
```
