```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# create a dataframe with sample data
df = pd.DataFrame({
    'Qubits': np.arange(10, 21 , 2),
    'Random Initialization': [-6.14, -3, -3.21, -2.19, -3.62 , -3.82 ],
    'CAFQA Initialization': [-59, -52, -38, -47, -19,  -10]
})

df.plot(x='Qubits', y=['Random Initialization', 'CAFQA Initialization'], kind='bar')
plt.xlabel('Qubits')
plt.ylabel('Initialization Values')
plt.title('Initialization for Random weighted complete graphs')
plt.show()
```

```python
# create a dataframe with sample data
df = pd.DataFrame({
    'Qubits': np.arange(10, 19 ,2 ),
    'Random Initialization': [-21, -32, -29, -36.97, -40 ],
    'CAFQA Initialization': [-29, -31, -62, -36, -32 ]
})

df.plot(x='Qubits', y=['Random Initialization', 'CAFQA Initialization'], kind='bar')
plt.xlabel('Qubits')
plt.ylabel('Initialization Values')
plt.title('Initialization for Random 3-regular graphs')
plt.show()
```
